# 00 · Start here: how these notebooks work

**Project:** burn footprint of the July 2026 fire complex in Ávila, Madrid and Toledo (Spain), made with Sentinel-2 satellite images.

### What is a notebook?
Think of it as a **lab notebook that can run code**. It is made of *cells*:

- **Text cells** (like this one) explain what is going on.
- **Code cells** contain Python. Run one with **Shift + Enter**. The result (a number, a table, a map) appears right underneath it.

Behind the notebook there is a **kernel**: the Python "engine" that runs your cells and *remembers* everything you created, until you restart it. That is why the order matters: if cell 5 uses a variable made in cell 3, run cell 3 first.

> In VS Code, check the top-right corner of the notebook. It should say **.venv (Python 3.14)**. If it doesn't, click it → *Python Environments* → **.venv**.

### The workflow, one notebook per step
| Notebook | What it does | Main outputs |
|---|---|---|
| 01_find_scenes | Finds a cloud-free image before and after the fire | `step1_scenes.json` |
| 02_reflectance | Downloads the bands and converts them to reflectance | `pre/post_reflectance.tif`, `valid_mask.tif` |
| 03_indices | Computes NDVI, NBR, BAI and GEMI, and their change | 12 index GeoTIFFs |
| 04_footprint | Thresholds + voting → burned / not burned → polygons | `burn_footprint.gpkg/.geojson/.kml` |
| 05_impact | Severity, vegetation classes, land cover burned | `severity_dNBR.tif`, `step5_impact.json` |
| 06_fire_weather | Weather on the ignition days vs the 1991-2020 climate | `step6_fire_weather.png` |
| 07_validation_effis | Compare the footprint with the official EFFIS perimeters | `step7_validation_map.png` |
| 08_progression | How the fire spread day by day, from VIIRS hotspots | `step8_progression.png` |

Each notebook **saves its results to disk** (folder `outputs_madrid_avila`), and the next notebook reads them. So you can close everything, come back tomorrow and continue from where you stopped.
